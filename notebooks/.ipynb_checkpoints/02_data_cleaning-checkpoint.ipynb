## 4. Diagnostic de la qualité des données

Cette section regroupe les principaux contrôles exploratoires nécessaires avant de définir les opérations de nettoyage.

### 4.1 Analyse des valeurs manquantes

Le taux de valeurs manquantes est calculé pour chaque variable. Cette mesure permet de distinguer les variables peu renseignées de celles dont l'information est presque complète.

**Point d'attention :** une valeur manquante ne signifie pas toujours une donnée inconnue. Dans le jeu House Prices, certaines absences peuvent représenter l'absence réelle d'un équipement ou d'une caractéristique. La stratégie de traitement doit donc tenir compte de la définition de chaque variable.

```python
def create_quality_summary(df: pd.DataFrame) -> pd.DataFrame:
    """Retourne un résumé des valeurs manquantes par variable."""

    missing_count = df.isna().sum()

    summary = pd.DataFrame({
        "Valeurs manquantes": missing_count,
        "Pourcentage manquant (%)": (
            missing_count / len(df) * 100
        ).round(2) if len(df) > 0 else 0
    })

    # Conserver uniquement les variables ayant des valeurs manquantes
    summary = summary[summary["Valeurs manquantes"] > 0]

    return summary.sort_values(
        by="Valeurs manquantes",
        ascending=False
    )


for name, df in dataframes.items():
    print(f"\nValeurs manquantes — {name.upper()}")

    summary = create_quality_summary(df)

    if not summary.empty:
        display(summary)
    else:
        print("Aucune valeur manquante détectée.")
```

### 4.2 Détection des valeurs potentiellement aberrantes

La méthode de l'écart interquartile (IQR) identifie les observations situées en dehors des bornes statistiques :

- borne inférieure : Q1 − 1,5 × IQR ;
- borne supérieure : Q3 + 1,5 × IQR ;
- IQR : Q3 − Q1.

Une observation détectée comme aberrante n'est pas nécessairement erronée. Elle doit être examinée au regard du contexte de la variable avant toute suppression ou transformation.

```python
def detect_outliers(df: pd.DataFrame) -> pd.DataFrame:
    """Détecte les valeurs aberrantes avec la méthode IQR."""

    results = []

    numeric_columns = df.select_dtypes(include="number").columns

    for col in numeric_columns:

        series = df[col].dropna()

        if series.empty:
            continue

        Q1 = series.quantile(0.25)
        Q3 = series.quantile(0.75)

        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = series[
            (series < lower_bound) |
            (series > upper_bound)
        ]

        results.append({
            "Variable": col,
            "Nombre de valeurs aberrantes": len(outliers),
            "Pourcentage (%)": round(
                len(outliers) / len(series) * 100, 2
            ),
            "Borne inférieure": lower_bound,
            "Borne supérieure": upper_bound
        })

    return pd.DataFrame(results)


for name, df in dataframes.items():

    print(f"\nDétection des valeurs aberrantes — {name.upper()}")

    outliers_summary = detect_outliers(df)

    if not outliers_summary.empty:
        display(
            outliers_summary.sort_values(
                by="Nombre de valeurs aberrantes",
                ascending=False
            )
        )
    else:
        print("Aucune variable numérique à analyser.")
```

### 4.3 Vérification de certaines incohérences

Les contrôles ci-dessous recherchent notamment :
- les chaînes de caractères vides ;
- les espaces superflus au début ou à la fin des chaînes ;
- les valeurs numériques infinies.

Ces vérifications constituent un premier diagnostic. Elles ne remplacent pas la validation métier des valeurs et des modalités.

```python
def detect_inconsistencies(df: pd.DataFrame) -> pd.DataFrame:
    """Détecte quelques incohérences courantes dans les données."""

    results = []

    for col in df.columns:

        # Vérification des chaînes vides et espaces inutiles
        if df[col].dtype == "object" or pd.api.types.is_string_dtype(df[col]):

            empty_values = df[col].apply(
                lambda x: isinstance(x, str) and x.strip() == ""
            ).sum()

            spaces_values = df[col].apply(
                lambda x: isinstance(x, str) and x != x.strip()
            ).sum()

            if empty_values > 0:
                results.append({
                    "Variable": col,
                    "Type d'incohérence": "Chaînes vides",
                    "Nombre": empty_values
                })

            if spaces_values > 0:
                results.append({
                    "Variable": col,
                    "Type d'incohérence": "Espaces inutiles",
                    "Nombre": spaces_values
                })

        # Vérification des valeurs infinies
        if pd.api.types.is_numeric_dtype(df[col]):

            infinite_values = np.isinf(
                df[col].dropna()
            ).sum()

            if infinite_values > 0:
                results.append({
                    "Variable": col,
                    "Type d'incohérence": "Valeurs infinies",
                    "Nombre": infinite_values
                })

    return pd.DataFrame(results)
```

```python
for name, df in dataframes.items():

    print(f"\nVérification des incohérences — {name.upper()}")

    inconsistencies = detect_inconsistencies(df)

    if not inconsistencies.empty:
        display(inconsistencies)
    else:
        print("Aucune incohérence détectée avec les contrôles effectués.")
```

## 5. Définition de la stratégie de traitement des valeurs manquantes

Le traitement doit être choisi **variable par variable**, en fonction de la signification de l'information et de son rôle dans le problème de prédiction.

### Principes de décision

| Situation observée | Traitement à envisager | Précaution |
|---|---|---|
| L'absence représente réellement une caractéristique inexistante | Créer une modalité explicite, par exemple `None` pour une variable catégorielle | Vérifier la définition de la variable |
| Variable numérique continue avec quelques valeurs inconnues | Imputation par médiane, éventuellement par groupe | Comparer la distribution avant/après |
| Variable catégorielle dont la valeur est inconnue | Modalité `Inconnue` ou imputation par le mode | Ne pas confondre inconnu et absence |
| Variable très fortement incomplète | Étudier son utilité et le mécanisme de non-réponse | Ne pas supprimer automatiquement |
| Valeur manquante dans la variable cible `SalePrice` | Ne pas imputer pour l'entraînement supervisé | Les lignes sans cible ne permettent pas d'apprendre la cible |

**Important :** les propositions ci-dessus sont des règles de décision générales. Elles doivent être validées à partir du dictionnaire des variables et des résultats du diagnostic avant d'être appliquées.

```python
# 5.1 Préparer un rapport détaillé des valeurs manquantes

def missing_values_report(df: pd.DataFrame) -> pd.DataFrame:
    """Retourne le nombre et le taux de valeurs manquantes par variable."""
    report = pd.DataFrame({
        "Nombre manquant": df.isna().sum(),
        "Taux manquant (%)": (df.isna().mean() * 100).round(2),
        "Type": df.dtypes.astype(str)
    })

    return (
        report.loc[report["Nombre manquant"] > 0]
        .sort_values("Taux manquant (%)", ascending=False)
    )


for dataset_name, dataset in dataframes.items():
    print(f"\nRapport des valeurs manquantes — {dataset_name.upper()}")
    report = missing_values_report(dataset)

    if report.empty:
        print("Aucune valeur manquante détectée.")
    else:
        display(report)


# 5.2 Définir explicitement les variables dont l'absence a une signification
# Compléter cette liste après consultation du dictionnaire des variables.
# Exemple : MISSING_MEANS_ABSENCE = ["PoolQC", "Alley", "Fence"]
MISSING_MEANS_ABSENCE = []

# Créer des copies de travail : les DataFrames sources restent inchangés.
df_train_clean = df_train.copy()
df_test_clean = df_test.copy()

# Appliquer uniquement les remplacements explicitement validés.
for column in MISSING_MEANS_ABSENCE:
    for dataset in (df_train_clean, df_test_clean):
        if column in dataset.columns:
            dataset[column] = dataset[column].fillna("None")

print("Copies de travail créées : df_train_clean et df_test_clean.")
print("Aucune imputation statistique automatique n'a été appliquée.")
print("Variables traitées comme une absence réelle :", MISSING_MEANS_ABSENCE)


# 5.3 Contrôle après traitement
for dataset_name, dataset in {
    "train": df_train_clean,
    "test": df_test_clean
}.items():
    remaining = int(dataset.isna().sum().sum())
    print(f"{dataset_name.upper()} — valeurs manquantes restantes : {remaining:,}")
```
